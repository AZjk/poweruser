### Initialize

```python

from pathlib import Path
from convert_nexus_to_csv import hdf2csv  # adjust import path if utils.py is elsewhere
```

### List all file names with the same header string for iterative conversion

```python


in_dir = Path("./Multitau")
out_dir = Path("./CSV")
out_dir.mkdir(parents=True, exist_ok=True)

prefix = "LargeG2-averaging"
exts = {".nxs", ".h5", ".hdf5", ".hdf"}

files = sorted(
    p for p in in_dir.iterdir()
    if p.is_file()
    and p.suffix.lower() in exts
    and p.name.startswith(prefix)
)
```

### Iterative conversion 

```python

ok = failed = 0
for p in files:
    try:
        hdf2csv(str(p), str(out_dir))
        ok += 1
    except Exception as e:
        print(f"[ERROR] {p.name}: {e}")
        failed += 1

print(f"\nDone. Converted {ok} file(s); {failed} failed. CSVs are in {out_dir.resolve()}")
```
